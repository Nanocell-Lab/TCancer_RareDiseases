# Etapa 2: Filtrar datos desde el proteoma humano

```python
import pandas as pd
# 
df = pd.read_csv("../data/Uniprot/uniprotkb_proteome_UP000005640_2023_11_21.tsv.gz",sep='\t',low_memory=False)
```

```python

print(df.shape)
print(df.columns)
print (df['Entry'].head(3))
```

```python
#aplicar filtro de proteinas transmembrana
df_transmembrane = df.dropna(subset=['Transmembrane'])
df_transmembrane.shape
```

**Codigo InterPRo**
# url
* IPR000333: Ser/Thr protein kinase, TGFB receptor
* IPR003605: GS domain

**Codigo Pfam**
# url
* PF07714: Protein tyrosine and serine/threonine kinase
* PF01064: Activin types I and II receptor domain

```python
input_interpro_code = ['IPR000333', 'IPR009030', 'IPR052112']

df_transmembrane = df_transmembrane.dropna(subset=['InterPro'])
df_tkl = df_transmembrane.loc[df_transmembrane['InterPro'].str.contains(interpro_code)].loc[df_transmembrane['Reviewed']=='reviewed']
#print(len(df_tkl['InterPro'].unique()))
#print(df_tkl.shape)
df_tkl.dropna(axis=1, inplace=True)
print("Total genes from {0} family: {1}".format(interpro_code, df_tkl.shape[0]))
print(df_tkl['Gene Names'].tolist())
df_tkl.head(3)
```

# Añadir ENSEMBL name

```python
tkl_names = [x[0] for x in df_tkl['Gene Names'].str.split(' ').tolist()]
df_tkl['Gen_names'] = tkl_names
df_tkl.head(2)
```

```python
from pyensembl import EnsemblRelease
# release 109/ Feb2023 uses human reference genome GRCh38
data = EnsemblRelease(109)
```

```python
ensembl_gene_name = []
for name in df_tkl['Gen_names']:
    ename = data.gene_ids_of_gene_name(name)
    ensembl_gene_name.append(ename[0])
print(len(ensembl_gene_name))
```

```python
# Agrega los nombres a la tabla
df_tkl['ENS ID'] = ensembl_gene_name
print(df_tkl.shape)
df_tkl[['Entry','ENS ID']].head(3)
```

```python
df_tkl.columns
```


# Añadir NCBI Gene ID

```python
import pandas as pd
dfncbi = pd.read_csv("../data/NCBI/NCBI_human_gene.tsv", sep='\t')
print(dfncbi.shape)
dfncbi.head(2)
```

```python
df_tkl_ncbi = df_tkl.merge(dfncbi, how='inner', on=None, left_on='Gen_names', right_on='Symbol')
```

```python
df_tkl_ncbi['NCBI ID'] = df_tkl_ncbi['GeneID_y']
df_tkl_ncbi['Uniprot Name'] = df_tkl_ncbi['Entry Name']
df_tkl_ncbi['Uniprot ID'] = df_tkl_ncbi['Entry']
print(df_tkl_ncbi.shape)
df_tkl_ncbi[['Gen_names','Uniprot Name','Aliases','Uniprot ID','NCBI ID','ENS ID']]
```
