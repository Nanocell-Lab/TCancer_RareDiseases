```python
gene_names = ['ACVR1', 'TGFBR2']  # Lista de nombres de genes

def get_exon_sequence(chromosome, start, end):
    # Método para obtener la secuencia de nucleótidos de un exón
    import requests
    # URL para la API de Ensembl
    url = f"https://rest.ensembl.org/sequence/region/homo_sapiens/{chromosome}:{start}-{end}?content-type=text/plain"
    # Realizar la solicitud
    response = requests.get(url)
    
    # Verificar si la solicitud fue exitosa
    if response.status_code == 200:
        return response.text.strip()  # Devolver la secuencia sin espacios en blanco
    else:
        print(f"Error al obtener la secuencia del exón: {response.status_code}")
        return None

def genename2exons(gene_name):
    # Método para obtener los exones de un transcrito canónico desde el nombre del gen
    import requests
    # URL para la API de Ensembl
    url = f"https://rest.ensembl.org/lookup/symbol/homo_sapiens/{gene_name}?expand=1"
    # Encabezados para la solicitud
    headers = {"Content-Type": "application/json"}
    # Realizar la solicitud
    response = requests.get(url, headers=headers)
    # Convertir la respuesta a formato JSON
    gene_info = response.json()

    # Verificar si la clave 'Transcript' está en los datos
    if 'Transcript' not in gene_info:
        print(f"No se encontraron transcritos para el gen {gene_name}")
        return []

    chromosome = gene_info.get('seq_region_name')
    canonical_transcript_id = gene_info.get('canonical_transcript')

    # Obtener y mostrar los exones
    exons = []  # Inicializar una lista vacía por si no se encuentra el transcrito canónico
    for transcript in gene_info['Transcript']:
        if transcript.get('is_canonical') == 1:
            exons = transcript['Exon']
            # Información adicional
            print(f"Lista de exones para: {gene_name}")
            print(f"Transcrito ID: {canonical_transcript_id}")
            print(f"Cromosoma: {chromosome}")
            print(transcript['assembly_name'])
            print(transcript['strand'])

    # Si no hay exones asociados, retornar una lista vacía
    if not exons:
        print(f"No se encontraron exones para el transcrito canónico del gen {gene_name}")
        return []

    exons_array = []            
    for i, exon in enumerate(exons):
        exon_sequence = get_exon_sequence(chromosome, exon['start'], exon['end'])  # Obtener la secuencia del exón
        if exon_sequence:
            exons_array.append((chromosome, i + 1, exon['start'], exon['end'], exon_sequence))  # Incluir la secuencia

    print(f"Cantidad de exones para {gene_name}: {len(exons_array)}")
    return exons_array

# Test para varios genes
for gene_name in gene_names:
    exons_array = genename2exons(gene_name)

    # Para un formato más bonito y separado por líneas
    for exon in exons_array:
        print(f"Exon {exon[1]}: {exon}")  # Usar exon[1] para mostrar el índice del exón correctamente
```

```python
gene_names = ['ACVR1', 'TGFBR2']

import requests

def get_sequence(region):
    """Método para obtener la secuencia de una región cromosómica"""
    url = f"https://rest.ensembl.org/sequence/region/homo_sapiens/{region}?content-type=text/plain"
    response = requests.get(url)
    if response.status_code == 200:
        return response.text.strip()
    else:
        print(f"Error al obtener la secuencia de la región {region}: {response.status_code}")
        return None

def get_transcript_sequences(gene_name):
    """Método para obtener las secuencias 5' UTR y 3' UTR de un gen"""
    url = f"https://rest.ensembl.org/lookup/symbol/homo_sapiens/{gene_name}?expand=1"
    headers = {"Content-Type": "application/json"}
    response = requests.get(url, headers=headers)
    
    if response.status_code != 200:
        print(f"Error en la solicitud para {gene_name}: {response.status_code}")
        return

    gene_info = response.json()

    if 'Transcript' not in gene_info:
        print(f"No se encontraron transcritos para el gen {gene_name}")
        return

    canonical_transcript_id = gene_info.get('canonical_transcript')

    for transcript in gene_info['Transcript']:
        if transcript['id'] == canonical_transcript_id:
            print(f"Transcrito canónico encontrado para {gene_name}: {canonical_transcript_id}")
            
            # Obtener coordenadas de UTRs si están disponibles
            utr_5_start = transcript.get('five_prime_utr', {}).get('start')
            utr_5_end = transcript.get('five_prime_utr', {}).get('end')
            utr_3_start = transcript.get('three_prime_utr', {}).get('start')
            utr_3_end = transcript.get('three_prime_utr', {}).get('end')

            # Obtener secuencias 5' UTR y 3' UTR si las coordenadas están presentes
            utr_5_seq = get_sequence(f"{transcript['seq_region_name']}:{utr_5_start}-{utr_5_end}") if utr_5_start and utr_5_end else None
            utr_3_seq = get_sequence(f"{transcript['seq_region_name']}:{utr_3_start}-{utr_3_end}") if utr_3_start and utr_3_end else None

            # Mostrar resultados o mensajes si no están disponibles
            if utr_5_seq:
                print(f"5' UTR para {gene_name}: {utr_5_seq}")
            else:
                print(f"5' UTR no disponible para {gene_name}")

            if utr_3_seq:
                print(f"3' UTR para {gene_name}: {utr_3_seq}")
            else:
                print(f"3' UTR no disponible para {gene_name}")
            
            break  # Salir después de obtener la secuencia del transcrito canónico

# Test para varios genes
for gene_name in gene_names:
    print(f"Secuencias UTR para el gen: {gene_name}")
    get_transcript_sequences(gene_name)
    print("\n")
```
