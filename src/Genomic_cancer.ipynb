```python
#DF mutaciones totales
import pandas as pd
family_name = 'genes_totales'
df_genes_totales = pd.read_csv('../data/mutaciones/df_frecuencias_mut_{0}.csv'.format(family_name))
print(df_genes_totales.shape)
print(df_genes_totales.columns)
df_genes_totales.head(2)
```

```python
df_genes_totales['verification_status'].unique()
```

```python
#DF de mutaciones en TGF-B
family_name = 'STKR'
df_tgfb = pd.read_csv('../data/mutaciones/df_count_mut_{0}.csv'.format(family_name))
print(df_tgfb.shape)
print(df_tgfb.columns)
df_tgfb.head(2)
```

```python
count_genes_tgfb = df_tgfb.groupby(['icgc_donor_id','Gen_names','tipo_cancer','aa_mutation'])['total_read_count'].sum().reset_index(name='total_read_count')
read_count_tgfb = count_genes_tgfb.groupby(['icgc_donor_id','Gen_names','tipo_cancer','total_read_count'])['aa_mutation'].nunique().reset_index(name='mutation_count')
read_count_tgfb = read_count_tgfb.groupby(['icgc_donor_id','Gen_names','tipo_cancer']).sum(numeric_only=True).reset_index()
read_count_tgfb.sort_values(by = 'mutation_count' )
```

```python
print(f"Numero total de mutaciones por paciente: {read_count_tgfb['mutation_count'].sum()}")
```

```python
#lista de pacientes con mutaciones en tgf-b
lista_tgfb = df_tgfb['icgc_donor_id'].unique()
```

```python
#Busca en DF de mut totales, las mutaciones correspondientes a todos los pacientes con mutaciones en TGF-B
merge_pacientes = df_genes_totales.loc[df_genes_totales['icgc_donor_id'].isin(lista_tgfb)]
merge_pacientes
```

```python
merge_pacientes['icgc_donor_id'].nunique()
```

# Procesar datos

## Contar mutaciones

```python
count_genes = merge_pacientes.groupby(['icgc_donor_id','gene_name','tipo_cancer','aa_mutation'])['total_read_count'].sum().reset_index(name='total_read_count')
read_count = count_genes.groupby(['icgc_donor_id','gene_name','tipo_cancer','total_read_count'])['aa_mutation'].nunique().reset_index(name='mutation_count')
read_count = read_count.groupby(['icgc_donor_id','gene_name','tipo_cancer']).sum(numeric_only=True).reset_index()
read_count.sort_values(by = 'mutation_count' )
```

```python
total_mut_pt = read_count['mutation_count'].sum()
total_mut_tgfb = read_count_tgfb['mutation_count'].sum()

print(f"Numero total de mutaciones por paciente afectado en tgf-b: {read_count['mutation_count'].sum()}")
print(f"Numero total de mutaciones en genes tgf-b: {read_count_tgfb['mutation_count'].sum()}")
print(f"Representatividad de mutaciones en tgf-b: {total_mut_tgfb/total_mut_pt}")
```

## Contar pacientes

```python
count_patients = read_count.groupby(['gene_name','tipo_cancer','total_read_count'])['icgc_donor_id'].nunique().reset_index(name='pacientes_count')
count_patients = count_patients.groupby(['gene_name','tipo_cancer']).sum(numeric_only=True).reset_index()
count_patients.sort_values(by = 'pacientes_count' )
```

## normalizar por secuenciacion

```python
# Calcular el total_read_count por paciente
read_count_pacientes = read_count.groupby(['icgc_donor_id'])['total_read_count'].sum(numeric_only=True).reset_index()
read_count_pacientes.rename(columns={'total_read_count': 'total_read_count_patient'}, inplace=True)

# Hacer un merge para unir los total_read_count por paciente al DataFrame original
read_count_genes_norm = read_count.merge(read_count_pacientes, on='icgc_donor_id', how='left')

read_count_genes_norm['total_read_patient(millon)'] = read_count_genes_norm['total_read_count_patient']/10**6

read_count_genes_norm['total_read_count(millon)'] = read_count_genes_norm['total_read_count']/10**6

# Calcular la columna norm_mutation_count
read_count_genes_norm['norm_mutation_count'] = read_count_genes_norm['mutation_count'] * (read_count_genes_norm['total_read_count(millon)'] / read_count_genes_norm['total_read_patient(millon)'])

#read_count_genes_norm['norm_mutation_count'] = read_count_genes_norm['mutation_count'] * (read_count_genes_norm['total_read_count'] / read_count_genes_norm['total_read_count_patient'])

# Mostrar el resultado
read_count_genes_norm.sort_values(by = 'mutation_count' )
```
