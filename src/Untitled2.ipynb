```python
import matplotlib.pyplot as plt

# Datos de ejemplo
uniprot_id = "P04637"
mutations = ["R273C", "R175H", "T125@5", "R248Q#7f3333@131"]
positions = [273, 175, 125, 248]  # Posiciones en la proteína para cada mutación

# Crear el gráfico
plt.figure(figsize=(10, 6))
plt.scatter(positions, [1] * len(positions), marker='o')

# Etiquetar cada mutación
for i, mutation in enumerate(mutations):
    plt.text(positions[i], 1.01, mutation, fontsize=12, ha='center')

plt.title(f"Mutaciones en la proteína p53 (UniProt ID: {uniprot_id})")
plt.yticks([])  # Quitar las etiquetas del eje y
plt.xlabel("Posición de la mutación en la proteína")
plt.xlim(100, 300)  # Ajustar límites del eje x según sea necesario
plt.grid()

# Mostrar el gráfico
plt.show()
```
