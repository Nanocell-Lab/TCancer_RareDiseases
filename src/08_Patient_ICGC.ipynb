```python
from lifelines.statistics import pairwise_logrank_test
import matplotlib.pyplot as plt
from lifelines import KaplanMeierFitter

df = concatenated_df

# Asignar las variables de tiempo y censura
T = df['Supervivencia (Meses)']
C = df['donor_vital_status'].map({'alive': True, 'deceased': False})

# Helper function for plotting Kaplan-Meier curves with counts and p-value
def plot_km_with_counts_and_pval(col):
    # Realizar Log-rank test y obtener el p-value
    filtered_df = df[df['Supervivencia (Meses)'] <= 60]
    log_rank = pairwise_logrank_test(
        filtered_df['Supervivencia (Meses)'], 
        filtered_df[col], 
        filtered_df['donor_vital_status']
    )
    p_value = log_rank.p_value.min()  # Obtener el menor p-value

    # Crear el gráfico Kaplan-Meier
    ax = plt.subplot(111)
    for r in df[col].unique():
        ix = df[col] == r
        count = ix.sum()  # Contar el número de pacientes para la categoría
        label = f"{r} (n={count})"  # Incluir el número de pacientes en la etiqueta
        kmf = KaplanMeierFitter()
        kmf.fit(T[ix], C[ix], label=label)
        kmf.plot(ci_show=False, ax=ax)
    
    ax.set_xlim(0, 60)  # Limitar el eje a 60 meses
    ax.set_title(f'Pacientes con mutaciones TGF-B y no TGF-B (60 meses)\np-value: {p_value:.4f}', weight='bold')
    ax.set_xlabel('Tiempo (Meses)')
    ax.set_ylabel('Probabilidad de supervivencia')
    plt.legend(loc='best')  # Ajustar la posición de la leyenda

# Helper function for printing out Log-rank test results
def print_logrank(col):
    filtered_df = df[df['Supervivencia (Meses)'] <= 60]
    log_rank = pairwise_logrank_test(
        filtered_df['Supervivencia (Meses)'], 
        filtered_df[col], 
        filtered_df['donor_vital_status']
    )
    return log_rank.summary

# Ejecución de funciones
print(print_logrank('Dataframe'))
plot_km_with_counts_and_pval('Dataframe')



#Guardar archivo
family_name = 'tgfb_no-tgfb'
plt.savefig("../run/imagenes/Kaplan-Meier_60meses_{0}.png".format(family_name), format='png', dpi=600)

# Ajustar diseño y mostrar
plt.tight_layout()
plt.show()
```

```python
from lifelines.statistics import pairwise_logrank_test
import matplotlib.pyplot as plt
from lifelines import KaplanMeierFitter

df = donor_survival

# Asignar las variables de tiempo y censura
T = df['Supervivencia (Meses)']
C = df['donor_vital_status'].map({'alive': True, 'deceased': False})


# Paleta de colores para cada categoría
color_palette = {
    '-60': '#ebac5d',  # Cambia 'Group1', 'Group2', etc., por los valores reales en tu columna
    '+60': '#733d00',
}

# Helper function for plotting Kaplan-Meier curves with counts and p-value
def plot_km_with_counts_and_pval(col):
    # Realizar Log-rank test y obtener el p-value
    filtered_df = df[df['Supervivencia (Meses)'] <= 60]
    log_rank = pairwise_logrank_test(
        filtered_df['Supervivencia (Meses)'], 
        filtered_df[col], 
        filtered_df['donor_vital_status']
    )
    p_value = log_rank.p_value.min()  # Obtener el menor p-value

    # Crear el gráfico Kaplan-Meier
    ax = plt.subplot(111)
    for r in df[col].unique():
        ix = df[col] == r
        count = ix.sum()  # Contar el número de pacientes para la categoría
        label = f"{r} (n={count})"  # Incluir el número de pacientes en la etiqueta
        kmf = KaplanMeierFitter()
        kmf.fit(T[ix], C[ix], label=label)
        
        # Usar el color correspondiente a la categoría
        kmf.plot(ci_show=False, ax=ax, color=color_palette.get(r, '#000000'))  # Color por categoría o negro por defecto

    ax.set_xlim(0, 60)  # Limitar el eje a 60 meses
    ax.set_title(f'Sobrevida de pacientes por rango de edad \np-value: {p_value:.4f}', 
                 fontsize=12, weight='bold')
    ax.set_xlabel('Tiempo (Meses)')
    ax.set_ylabel('Probabilidad de supervivencia')
    plt.legend(loc='best')  # Ajustar la posición de la leyenda

# Helper function for printing out Log-rank test results
def print_logrank(col):
    filtered_df = df[df['Supervivencia (Meses)'] <= 60]
    log_rank = pairwise_logrank_test(
        filtered_df['Supervivencia (Meses)'], 
        filtered_df[col], 
        filtered_df['donor_vital_status']
    )
    return log_rank.summary

# Calcular el porcentaje de supervivencia global por categoría
def calculate_survival_percentages(categories, time_point=60):
    survival_percentages = {}
    for category in categories:
        ix = df['age_group'] == category
        kmf = KaplanMeierFitter()
        kmf.fit(T[ix], C[ix], label=category)
        # Obtener la probabilidad de supervivencia en el punto de tiempo especificado
        survival_percentages[category] = kmf.survival_function_at_times(time_point).values[0] * 100
    return survival_percentages

# Categorías y cálculo
categories = ['-60', '+60']  # Cambia según las categorías reales en tu columna
survival_percentages = calculate_survival_percentages(categories, time_point=60)

# Mostrar los resultados
for category, survival in survival_percentages.items():
    print(f"Porcentaje de supervivencia a los 60 meses para {category}: {survival:.2f}%")

# Opcional: puedes usar estas tasas en tu gráfico Kaplan-Meier para anotarlas.


# Ajustar Kaplan-Meier para toda la cohorte y calcular la mediana
kmf_global = KaplanMeierFitter()
kmf_global.fit(T, C, label='Global Survival')

# Mostrar la mediana de supervivencia global
print(f"Mediana de supervivencia global: {kmf_global.median_survival_time_}")

# Ejecución de funciones
print(print_logrank('age_group'))
plot_km_with_counts_and_pval('age_group')

plt.tight_layout()

# Guardar archivo
family_name = 'tgfb'
plt.savefig("../run/imagenes/tesis/donor_age_{0}.svg".format(family_name), format='svg')

plt.show()
```
