```python

# df_tkl es una lista de genes
genes_tkl = ['BMPR1B', 'ACVR2A', 'BMPR1A', 'ACVR1B', 'TGFBR1', 'ACVRL1', 'TGFBR2', 'ACVR1', 'ACVR2B', 'BMPR2', 'AMHR2', 'ACVR1C']

# Selecciona las filas donde la columna 'Name' contiene alguna de las cadenas en genes_tkl
df = df_concat_genes[df_concat_genes['Name'].str.contains('|'.join(genes_tkl))]

# Ahora df_concat_genes contiene las filas que cumplen con esa condición
print(df.shape)
df.head()
```

```python
sorted(df['Gene(s)'].unique())
```

```python
#Cambiar el nombre de las filas
import pandas as pd
import re

# Lista de genes predefinida
lista_genes = ['ACVR1', 'ACVR1B', 'ACVR1C', 'ACVR2A', 'ACVR2B', 'ACVRL1', 
               'AMHR2', 'BMPR1A', 'BMPR1B', 'BMPR2', 'TGFBR1', 'TGFBR2']


# Definir función para extraer el nombre del gen principal
def extract_gene_name(gene_string):
    genes = gene_string.split('|')  # Dividir la cadena por '|'
    for gene in genes:
        if not gene.startswith('LOC'):  # Si el gen no comienza con 'LOC', es el principal
            return gene
    return None  # Devolver None si no se encuentra un gen principal válido

# Aplicar la función a la columna Gene(s)
df['Gene(s)'] = df['Gene(s)'].apply(extract_gene_name)

# Filtrar y mostrar solo los genes que están en la lista predefinida
df_concat_genes = df[df['Gene(s)'].isin(lista_genes)]

# Mostrar resultado final
print("Genes extraídos y verificados:")
print(df_concat_genes['Gene(s)'].unique())
print("Tamaño del DataFrame después del filtrado:", df_concat_genes.shape)
```

```python
import altair as alt
chart1=alt.Chart(df_concat_genes
).mark_bar().encode(
    y=alt.Y('count(Protein change):Q', title='Numero de mutaciones'),
    x=alt.X('Gene(s)', sort='-y', title='Tipo de cancer',axis=alt.Axis(
              labelAngle=40, labelFontSize=12)),
).properties(
    title="Gen con mayor numero de mutaciones patogenicas"
)
alt.data_transformers.enable('default', max_rows=None)
chart2=alt.Chart(df_concat_genes).mark_bar().encode(
    x=alt.X('Condition(s)', sort='-y', title='Tipo de cancer',axis=alt.Axis(
              labelAngle=40, labelFontSize=12)),
    y='count(Protein change):Q',
    color=alt.Color('Gene(s)', scale=alt.Scale(scheme='category20'))
)
chart1|chart2
```

```python
#Elimina las mutaciones sinonimas
df_concat_genes.dropna(subset='Molecular consequence', inplace=True)
df_concat_genes=df_concat_genes[~df_concat_genes['Molecular consequence'].str.contains('synonymous', case=False, na=False)]
df_concat_genes.shape
```

```python
# Utilizamos la función explode() para crear una nueva fila para cada elemento en la lista
# Separar las enfermedades y expandir en filas individuales
df_concat_genes['Condition(s)'] = df_concat_genes['Condition(s)'].astype(str)

# Separar las enfermedades y expandir en filas individuales
df_concat_genes['Condition(s)'] = df_concat_genes['Condition(s)'].str.split('|')
df_clinvar = df_concat_genes.explode('Condition(s)')

# Ahora df_clinvar tiene una fila separada para cada enfermedad
df_clinvar.shape
```

```python
# Dividir las mutaciones separadas por comas en listas
df_clinvar['Protein change'] = df_clinvar['Protein change'].str.split(', ')

# Usar explode para convertir las listas en filas individuales
df_clinvar = df_clinvar.explode('Protein change').reset_index(drop=True)

# Mostrar el DataFrame transformado
print("\nDataFrame transformado:")
print(df_clinvar.shape)
```

```python
import pandas as pd

# Crear un diccionario con los DataFrames por gen
genes = ['ACVR1', 'ACVR1B', 'ACVR1C', 'ACVR2A', 'ACVR2B', 'BMPR1A', 'BMPR1B', 'BMPR2', 'TGFBR1', 'TGFBR2', 'AMHR2', 'ACVRL1']
gene_dfs = {gene: df_merge_genes[df_merge_genes['TK_Name'] == gene] for gene in genes}

# Función para contar mutaciones por gen
def contar_mutaciones(df):
    return df.groupby(['tipo_cancer', 'cancer_name', 'clasificacion', 'Condition(s)', 'Gene(s)','VariationID'])['aa_mutation'].nunique().reset_index(name='num_mutaciones')

# Función para detallar mutaciones por gen
def detallar_mutaciones(df):
    return df.groupby(['tipo_cancer', 'cancer_name', 'clasificacion', 'Condition(s)', 'Gene(s)','VariationID','Germline classification'])['aa_mutation'].apply(list).reset_index(name='mutaciones')

# Crear diccionarios para almacenar los resultados
mutaciones_por_gen = {}
detalles_por_gen = {}

# Aplicar las funciones a cada DataFrame en el diccionario
for gene, df in gene_dfs.items():
    mutaciones_por_gen[gene] = contar_mutaciones(df)
    detalles_por_gen[gene] = detallar_mutaciones(df)

# Mostrar los resultados
for gene in genes:
    print(f"{gene} - Número de mutaciones por gen y tipo de cáncer:")
    print(mutaciones_por_gen[gene])
    print(f"\n{gene} - Detalle de mutaciones:")
    print(detalles_por_gen[gene])
    print("\n" + "="*80 + "\n")
```

```python
import pandas as pd

# Crear un diccionario con los DataFrames por gen
genes = ['ACVR1', 'ACVR1B', 'ACVR1C', 'ACVR2A', 'ACVR2B', 'BMPR1A', 'BMPR1B', 'BMPR2', 'TGFBR1', 'TGFBR2', 'AMHR2', 'ACVRL1']
gene_dfs = {gene: df_merge_genes[df_merge_genes['TK_Name'] == gene] for gene in genes}

# Función para detallar mutaciones por gen
def detallar_mutaciones(df):
    return df.groupby(['tipo_cancer', 'cancer_name', 'Condition(s)','Germline classification', 'Gene(s)', 'clasificacion'])['aa_mutation'].apply(list).reset_index(name='mutaciones')

# Crear diccionario para almacenar los detalles
detalles_por_gen = {}

# Aplicar la función a cada DataFrame en el diccionario
for gene, df in gene_dfs.items():
    detalles_por_gen[gene] = detallar_mutaciones(df)

# Guardar en un solo archivo Excel con hojas separadas para cada gen
with pd.ExcelWriter('../run/Mutaciones_gen_crossroad/detalles_mutaciones_genes.xlsx') as writer:
    for gene, df in detalles_por_gen.items():
        df.to_excel(writer, sheet_name=gene, index=False)

print("Datos guardados en '../run/Mutaciones_gen_crossroad/detalles_mutaciones_genes.xlsx'.")
```
