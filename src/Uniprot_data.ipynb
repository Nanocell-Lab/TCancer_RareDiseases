```python
import requests
import pandas as pd
```

```python
#Nombre grupo de genes
grupo = "kit"
#Insetar código de uniprot
uniprot_ids = ["P10721", "P16234", "P09619", "P36888", "P07333"]
```

```python

#Función para obtener UniprotID

data = []  

for uniprot_id in uniprot_ids: 
    response = requests.get(f'https://rest.uniprot.org/uniprotkb/{uniprot_id}.json')

    if response.status_code == 200:
        json_data = response.json()
        gen_name = json_data['genes'][0]['geneName']['value']  # Extrae el nombre del gen
        
        # Guarda los datos en la lista
        data.append({"uniprot_id": uniprot_id, "gen_name": gen_name}) 
    else:
        print(f"Error al obtener {uniprot_id}: {response.status_code}")

df = pd.DataFrame(data)

# DataFrame con genes y ID's

df
```

```python
name = ['ACVR1']

data = []  


for gen in name:

    
    url = "https://rest.uniprot.org/uniprotkb/search"
    params = {
      "query": f"gene_exact:{gen} AND reviewed:true AND Homo sapiens"}
    response = requests.get(url, params=params)
    json_data = response.json()

    gen_name = json_data.get("results")[0].get("genes")[0].get("geneName").get("value")
    uniprot_id = json_data.get("results")[0].get("primaryAccession")
    
    if gen_name == gen:
        # Guarda los datos en la lista
        data.append({"uniprot_id": uniprot_id, "gen_name": gen_name}) 

df = pd.DataFrame(data)

# DataFrame con genes y ID's

df

```

```python
import requests, sys, json

params = {
  "query": "gene_field AND reviewed:true",
  "fields": [
    "accession",
    "protein_name",
    "cc_function",
    "ft_binding"
  ],
  "sort": "accession desc",
  "size": "50"
}
headers = {
  "accept": "application/json"
}
base_url = "https://rest.uniprot.org/uniprotkb/search"

response = requests.get(base_url, headers=headers, params=params)
if not response.ok:
  response.raise_for_status()
  sys.exit()

data = response.json()

print(json.dumps(data, indent=2))
```
